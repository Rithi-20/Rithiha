```python
from sentence_transformers import SentenceTransformer
import chromadb
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
print("Imports Successful")
```

```python
document = """
Data Analytics Professional Program

The Data Analytics Professional Program is a 12-week training program
designed to help learners develop practical skills in data analytics.

The program includes Python, SQL, Statistics, Advanced Excel, Power BI,
and Generative AI.

The program is delivered through instructor-led sessions, self-learning
activities, assessments, and practical projects.

Students are expected to maintain at least 80% attendance throughout
the program.

The program includes three major practical milestones. Each milestone
requires students to apply the concepts learned during the training.

The final capstone project requires students to work on an end-to-end
business problem using data analytics techniques.

Students are expected to submit project work within the deadlines
communicated by the training team.

The Python module covers Python fundamentals, NumPy, Pandas, data
manipulation, data analysis, and visualization.

The SQL module covers relational databases, SQL queries, filtering,
aggregation, joins, subqueries, and practical business scenarios.

The Statistics module covers descriptive statistics, probability,
sampling, distributions, correlation, confidence intervals, and
hypothesis testing.

The Power BI module covers data loading, transformation, data modeling,
DAX fundamentals, visualizations, dashboards, and report publishing.

The Generative AI module introduces large language models, prompt
engineering, embeddings, vector databases, retrieval-augmented generation,
and AI application development.

The final project should demonstrate the student's ability to understand
a business problem, prepare data, analyze the data, communicate insights,
and use appropriate AI techniques where relevant.
"""
```

```python
print(document)
```

```python
paragraphs = [
    p.strip()
    for p in document.split("\n\n")
    if p.strip()
]

chunks = []

for i, paragraph in enumerate(paragraphs):
    chunks.append({
        "id": f"chunk_{i}",
        "text": paragraph
    })

print(f"Number of chunks: {len(chunks)}")

for chunk in chunks:
    print("\n", chunk["id"])
    print(chunk["text"])
```

```python
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded.")
```

```python
texts = [chunk["text"] for chunk in chunks]

embeddings = embedding_model.encode(
    texts,
    normalize_embeddings=True
)

print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape[1])
```

```python
chunk_index = 5

print("CHUNK:")
print(chunks[chunk_index]["text"])

print("\nVECTOR:")
print(embeddings[chunk_index])
```

```python
chunk_index = 0

print("CHUNK:")
print(chunks[chunk_index]["text"])

print("\nVECTOR:")
print(embeddings[chunk_index])
```

```python
test_sentences = [
    "Students must maintain 80 percent attendance.",
    "Learners are required to attend at least eighty percent of classes.",
    "The program teaches SQL and Python."
]

test_embeddings = embedding_model.encode(
    test_sentences,
    normalize_embeddings=True
)

similarity_matrix = test_embeddings @ test_embeddings.T

print("Similarity matrix:\n")
print(similarity_matrix)
```

```python
client = chromadb.Client()

collection = client.get_or_create_collection(
    name="data_analytics_program"
)

print("Chroma collection created.")
```

```python
collection.add(
    ids=[chunk["id"] for chunk in chunks],
    documents=texts,
    embeddings=embeddings.tolist()
)

print(f"Stored {len(chunks)} chunks in Chroma.")
```

```python
question = "What is the attendance requirement for students?"

print("Question:", question)
```

```python
query_embedding = embedding_model.encode(
    [question],
    normalize_embeddings=True
)

print("Query embedding shape:", query_embedding.shape)
```

```python
results = collection.query(
    query_embeddings=query_embedding.tolist(),
    n_results=3
)

retrieved_ids = results["ids"][0]
retrieved_documents = results["documents"][0]

print("Retrieved chunks:\n")

for rank, (chunk_id, doc) in enumerate(
    zip(retrieved_ids, retrieved_documents), start=1
):
    print(f"\n--- Result {rank} | {chunk_id} ---")
    print(doc)
```

```python
context = "\n\n".join(retrieved_documents)

print("Retrieved Context:\n")
print(context)
```

```python
model_name = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

generator = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto"
)

print("Generator model loaded.")
```

```python
prompt = f"""
You are a helpful assistant.

Answer the user's question using only the information provided
in the context below.

If the answer is not present in the context, say:
"I don't have enough information in the provided documents."

Context:
{context}

Question:
{question}

Answer:
"""

print(prompt)
```

```python
messages = [{"role": "user", "content": prompt}]
chat_prompt = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(
    chat_prompt,
    return_tensors="pt"
).to(generator.device)

with torch.no_grad():
    outputs = generator.generate(
        **inputs,
        max_new_tokens=100,
        temperature=0.2,
        do_sample=True
    )

gen_ids = outputs[0][inputs["input_ids"].shape[1]:]
answer = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

print("Answer:")
print(answer)
```

```python
def rag_chat(question, top_k=3):

    # 1. Embed the question
    query_embedding = embedding_model.encode(
        [question],
        normalize_embeddings=True
    )

    # 2. Retrieve relevant chunks
    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=top_k
    )

    retrieved_documents = results["documents"][0]

    # 3. Build context
    context = "\n\n".join(retrieved_documents)

    # 4. Build prompt
    prompt = f"""
You are a helpful assistant.

Answer the question using only the provided context.

If the answer is not present in the context, say:
"I don't have enough information in the provided documents."

Context:
{context}

Question:
{question}

Answer:
"""

    # 5. Generate answer (chat template + token-based slicing)
    messages = [{"role": "user", "content": prompt}]
    chat_prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        chat_prompt,
        return_tensors="pt"
    ).to(generator.device)

    with torch.no_grad():
        outputs = generator.generate(
            **inputs,
            max_new_tokens=100,
            temperature=0.2,
            do_sample=True
        )

    gen_ids = outputs[0][inputs["input_ids"].shape[1]:]
    answer = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

    return answer, retrieved_documents
```

```python
answer, sources = rag_chat(
    "What topics are covered in the Python module?"
)

print("Answer:\n")
print(answer)

print("\n\nRetrieved Sources:\n")

for i, source in enumerate(sources, start=1):
    print(f"--- Source {i} ---")
    print(source)
```
