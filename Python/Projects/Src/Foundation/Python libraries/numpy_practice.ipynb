Multidimensional array(2d)

```python
mark=np.array([
    [90,90,78],
    [78,56,98],
    [98,67,89]]
)
print(mark)

#indexing
print(mark[0][0])  # or 
print(mark[0,0])

# Slicing & indexing
print(mark[0:3])  
print(mark[0:2][0:2])


print(mark[1:,0])   # ------> 1 represent row and 0 represent column now 1: -> access 78,56,98  ,98,67,89 and in that 0 th column so output is 78,98
print(mark[1:][0])  #1: -> access 78,56,98  ,98,67,89  in that 0 th row so 78,56,98 is output

print(mark[0][0:2])
print(mark[0,0:2])
```

Axis

```python
b1=np.array([
    [90,90,78],
    [78,56,98]])
b2=np.array([
    [66,70,88],
    [46,56,33]])

comb=np.concatenate((b1,b2),axis=0) # axis =0 represent rows(x axis)
# [[90 90 78]
#  [78 56 98]
#  [66 70 88]
#  [46 56 33]]
print(comb)

comb1=np.concatenate((b1,b2),axis=1) # axis =1 represent columns(y axis)
print(comb1)
# [[90 90 78 66 70 88]
#  [78 56 98 46 56 33]]
```

To search a value - where

```python
b1=np.array([
    [90,90,78],   
    [78,56,98]])
#Find where the mark is 78
print(np.where(b1==78))

# print marks greater than 78
print(np.where(b1>78))     # o/p: (array([0, 0, 1]), array([0, 1, 2])) --> 0,0=90   0,1=90    1,2=98

# instead of position print numbers
print(b1[np.where(b1>78)])  
```

To sort an array

```python
b1=np.array([
    [90,90,78],   
    [78,56,98]])

print(np.sort(b1))  # sort in ascending order by row

print(np.sort(b1)[::-1])  # to take the complete array and to sort it 

print(np.sort(b1,axis=0))  # to sort row wise

print(np.sort(b1,axis=1))  # to sort column wise
```

A shallow copy creates a new array that refers to the same data as the original
array, while a deep copy creates a completely new array with its own data.
Modifying a deep copy does not affect the original array.


You can create a shallow copy of a NumPy array using the numpy.view() method. 
You can create a deep copy of a NumPy array using the numpy.copy() function

You can use numpy.loadtxt() to read data from a text file into a NumPy array.

numpy.save() saves data in a binary format, which is more efficient and
preserves data types and structures. numpy.savetxt() saves data as plain text,
which is human-readable but less efficient.

numpy.savez() allows you to save multiple NumPy arrays into a single
