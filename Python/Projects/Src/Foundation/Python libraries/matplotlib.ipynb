```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105,106,107],
    "Name": ["Arun", "Priya", "Karthik", "Meena", "Rahul","Rithi","Nisha"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT",None,None],
    "Attendance": [82, 91, 76, 68, 88, 90, 89],
    "Python": [85, 92, 76, 65, 88, None, 90],
    "SQL": [78, 88, 82, 70, 91,45,34],
    "Excel": [90, 95, 80, 72, 86,45,78]
}

df=pd.DataFrame(data) # Dataframe
print(df)
```

```python
dep_by_avg=df.groupby("Department")["Python"].mean()
dep_by_avg.plot(kind='bar',color='red')
plt.title("Department by average")
plt.xlabel("Department")
plt.ylabel("Average")
plt.figure(figsize=(5,6))  # width - 5, height -6
plt.show()
```

```python
months=["Jan","Feb","Mar","Apr","May","Jun"]
att=[45,67,78,45,89,90]

plt.plot(months,att,c='purple',marker='.',linewidth=5, linestyle='--')  # by default line chart

# colormap represent combination of colors
```

```python
# Histogram - it is used for distribution
plt.hist(df["Python"],bins=5)
plt.title("Histogram")
```

```python
df["Average"]=(df["Python"]+df["SQL"]+df["Excel"])//3
df["Results"]=df["Average"].apply(lambda x: "Pass" if x>=50 else "Fail")
```

```python
plt.scatter(df["Attendance"],df["Average"])
plt.xlabel("Attendance")
plt.ylabel("Average")
plt.title("Attendance by average marks")
```

A subplot is a way to create multiple plots within a single figure. It's useful in data
visualization when you want to display multiple related plots side by side or in a
grid to compare and analyze data more effectively.

You can customize the layout of subplots by adjusting the number of rows and
columns in the grid (e.g., plt.subplots(2, 2)), and you can set titles, labels, and
other attributes for each subplot individually using the axs objects. Additionally,
you can adjust the spacing between subplots using plt.subplots_adjust() and
other layout parameters
