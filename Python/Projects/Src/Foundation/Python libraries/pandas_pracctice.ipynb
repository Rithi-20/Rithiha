Filtering

```python
# # student rows whose rows > 75
cond=df[df["Attendance"]>=75]
print(cond.Name)

# Find students whose attendance is atleast 75 and python mark is atleast 80
print(df.loc[(df["Attendance"]>=75) & (df["Python"]>=80)])

# Calculate the avg of python,sql and excel
# print((df[["Python","SQL","Excel"]].sum())/3)

# Calculate the avg of python,sql and excel for each person
(df["Python"]+df["SQL"]+df["Excel"])//3
```

Data Manipulation

```python
# To add a column
df["Average"]=(df["Python"]+df["SQL"]+df["Excel"])//3
df
```

```python
# Add a column when the average is greater than 50 pass else fail
df["Results"]=df["Average"].apply(lambda x: "Pass" if x>=50 else "Fail")
df
```

Sort

```python
# Ascending order
df.sort_values(by="Average") # ascending default

# Descending order
df.sort_values(by="Average",ascending=False)
```

Group by

```python
# what is the avg python mark for each department
df.groupby("Department")["Python"].mean()
```

```python
# Number of students per department
df.groupby("Department")["Student_ID"].count()
```

```python
# Highest python mark by department
df.groupby("Department")["Python"].max()
```

Group by with multiple metrics

```python
df.groupby("Department").agg({
    "Student_ID":"count",
    "Python":"mean",
    "Excel":"mean",
    "SQL":"mean",
    "Attendance":"mean"
    })
```
